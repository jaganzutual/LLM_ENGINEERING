```python
from transformers import BertModel, AutoTokenizer
import pandas as pd

```

```python
model_name = "bert-base-uncased"
model = BertModel.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
sentance = "When life gives you lemons, make lemonade."
```

```python
tokens = tokenizer.tokenize(sentance)
print(tokens)
```

```python
new_token_ids = tokenizer.encode(sentance)
print(new_token_ids)
```

```python
text = "Tokenize me this please"
encoded_input = tokenizer(text, return_tensors='pt') #pr stance for PyTorch and TensorFlow use 'tf' and np for NumPy
print(encoded_input)
```

```python
output = model(**encoded_input)
```

```python
print(output.last_hidden_state)
print(output.last_hidden_state.shape)
```
