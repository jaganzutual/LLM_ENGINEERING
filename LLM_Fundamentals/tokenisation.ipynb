```python
import torch
from transformers import AutoTokenizer, AutoModel
```

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased")

```

```python
sentence1 = "she took a rose from the garden"
sentence2 = "Sun Rose in the Morning"
```

```python
inputs1 = tokenizer(sentence1, return_tensors="pt",padding=True)["input_ids"]
print(inputs1)
inputs2 = tokenizer(sentence2, return_tensors="pt")["input_ids"]
print(inputs2)
```

```python
with torch.no_grad():
    outputs1 = model(inputs1).last_hidden_state
    outputs2 = model(inputs2).last_hidden_state

print(outputs1.shape)
print(outputs2)
```
