```python
import torch
import torch.nn as nn
import math

torch.manual_seed(42)

print("✅ Ready for Day 2: Multi-Head Attention!")
print("\nWe'll go one concept at a time.")
```

```python
"""
Why do we need Multi-Head Attention?
Understanding the limitation of single-head
"""

print("🤔 THE PROBLEM:")
print("="*60)

print("\nYesterday we built SINGLE-HEAD attention:")
print("  - One attention mechanism")
print("  - Can only capture ONE type of pattern")

print("\nExample sentence: 'The cat sat'")
print("\nSingle-head might focus on:")
print("  - 'cat' → 'sat' (subject-verb relationship)")

print("\nBUT what about other relationships?")
print("  - 'The' → 'cat' (determiner-noun)")
print("  - 'cat' → 'The' (backward reference)")
print("  - Position information")

print("\n❌ Single head can't capture ALL relationships at once!")
print("\n✅ SOLUTION: Use MULTIPLE heads (each learns different patterns)")

print("\n" + "="*60)
```

```python
"""
What is Multi-Head Attention?
Simple concept - no math yet!
"""

print("💡 MULTI-HEAD ATTENTION SOLUTION:")
print("="*60)

print("\nInstead of ONE attention mechanism...")
print("We use MULTIPLE attention mechanisms in PARALLEL!")

print("\nExample with 4 heads:")
print("  Head 1: Learns subject-verb relationships")
print("  Head 2: Learns adjective-noun relationships")
print("  Head 3: Learns positional patterns")
print("  Head 4: Learns semantic similarity")

print("\n🔄 Each head runs INDEPENDENTLY")
print("   They all work at the SAME TIME (parallel)")

print("\n📦 Then we COMBINE all their outputs")
print("   Final output has information from ALL heads!")

print("\n" + "="*60)

print("\n🎯 Key Idea:")
print("   Multiple heads = Multiple perspectives")
print("   Like having 4 experts instead of 1!")
```

```python
"""
How do we split dimensions across multiple heads?
Simple example with numbers - using full names!
"""

print("🔢 SPLITTING DIMENSIONS ACROSS HEADS:")
print("="*60)

# Simple example
embedding_dimension = 64      # Total dimensions we have
number_of_heads = 4          # We want 4 heads

# Calculate dimensions per head
dimension_per_head = embedding_dimension // number_of_heads

print(f"\nTotal dimensions available: {embedding_dimension}")
print(f"Number of heads we want: {number_of_heads}")
print(f"Dimensions per head: {embedding_dimension} ÷ {number_of_heads} = {dimension_per_head}")

print(f"\n📊 How it splits:")
print(f"  Head 1 works with: {dimension_per_head} dimensions")
print(f"  Head 2 works with: {dimension_per_head} dimensions")
print(f"  Head 3 works with: {dimension_per_head} dimensions")
print(f"  Head 4 works with: {dimension_per_head} dimensions")
print(f"  ─────────────────────────────────────────────")
print(f"  Total: {number_of_heads} × {dimension_per_head} = {embedding_dimension} dimensions ✅")

print("\n💡 Each head gets its own slice of the dimensions!")
print("   Like dividing a pizza into 4 equal slices 🍕")

print("\n" + "="*60)
```

```python
"""
Each head has its own set of Q, K, V weights
All heads see the same input, but create different projections
"""

print("🔑 EACH HEAD HAS ITS OWN WEIGHTS:")
print("="*60)

embedding_dimension = 64
number_of_heads = 4
dimension_per_head = 16

print(f"\nInput: embedding [{embedding_dimension} dimensions]")
print("\nEach head has 3 weight matrices:")

for head_num in range(1, number_of_heads + 1):
    print(f"\nHead {head_num}:")
    print(f"  weight_query_head{head_num}:  [{embedding_dimension}, {dimension_per_head}]")
    print(f"  weight_key_head{head_num}:    [{embedding_dimension}, {dimension_per_head}]")
    print(f"  weight_value_head{head_num}:  [{embedding_dimension}, {dimension_per_head}]")

print("\n" + "="*60)

print("\n💡 What happens:")
print(f"  1. All heads receive SAME input: [{embedding_dimension} dims]")
print(f"  2. Each head projects using ITS OWN weights")
print(f"  3. Each head gets: [{dimension_per_head} dims]")
print(f"  4. Different weights = different information extracted!")

print("\n🎯 Example:")
print("  Head 1 might learn to extract 'subject' information")
print("  Head 2 might learn to extract 'verb' information")
print("  Head 3 might learn to extract 'adjective' information")
print("  Head 4 might learn to extract 'position' information")
```

```python
"""
Step 1: Create the class and initialize basic parameters
Using our familiar variable names!
"""

class MultiHeadAttention(nn.Module):
    def __init__(self, embedding_dimension, number_of_heads):
        """
        Args:
            embedding_dimension: Total embedding dimension (e.g., 64)
            number_of_heads: Number of attention heads (e.g., 4)
        """
        super().__init__()
        
        # Validation: embedding_dimension must be divisible by number_of_heads
        assert embedding_dimension % number_of_heads == 0, "embedding_dimension must be divisible by number_of_heads"
        
        # Store the parameters
        self.embedding_dimension = embedding_dimension
        self.number_of_heads = number_of_heads
        self.dimension_per_head = embedding_dimension // number_of_heads
        
        print("✅ MultiHeadAttention class created!")
        print(f"   Total dimensions: {self.embedding_dimension}")
        print(f"   Number of heads: {self.number_of_heads}")
        print(f"   Dimensions per head: {self.dimension_per_head}")

# Test: Create an instance
mha = MultiHeadAttention(embedding_dimension=64, number_of_heads=4)
```
